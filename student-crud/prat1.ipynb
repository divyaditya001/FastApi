```python
import random
import string
```

```python
string.ascii_letters, string.digits, string.punctuation
```

```python
# all_digits = string.ascii_letters + string.digits + string.punctuation
all_digits = string.ascii_letters + string.digits 
# print(all_digits)
print(''.join(random.choices(all_digits, k=10)))
```

```python
import secrets
print(''.join(secrets.choice(all_digits) for _ in range(10)))
```

```python
def password(len):
  all_digits = string.ascii_letters + string.digits + string.punctuation
  password = ''.join(secrets.choice(all_digits) for _ in range(len))
  return password
```
